Importing the dependencies

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

#Import the model
from model import LogisticRegression
```

Data Collection and Analysis

PIMA Diabetes dataset

```python
#loading the dataset in a pandas data frame
diabetes_dataset = pd.read_csv('../Logistic_Regression/dataset/diabetes.csv')
```

```python
diabetes_dataset.head()
```

```python
#Dimensions of dataset
diabetes_dataset.shape
```

```python
#Getting the statistical measures
diabetes_dataset.describe()
```

```python
#Getting the data distribution
diabetes_dataset['Outcome'].value_counts()
```

```python
#group by outcome, select the aggregate measures
diabetes_dataset.groupby('Outcome').mean()
```

```python
#Separating the data and labels
X = diabetes_dataset.drop('Outcome', axis = 1)
Y = diabetes_dataset['Outcome']
```

```python
print(X)
```

```python
print(Y)
```

**Split the data in training and testing**

```python
#stratify -> keeps the ratio of outcome 0:1 same for original, test data and training data
train_X, test_X, train_Y, test_Y = train_test_split(X, Y, test_size = 0.3, stratify = Y, random_state = 2)
```

**Data Standardization**

```python
#Call the constructor StandardScaler()
scaler = StandardScaler()
```

```python
#Fit the mean and standard deviation **for each column**
scaler.fit(train_X)
```

```python
#Transform training and testing data using the measures
#of training data only to avoid data leakage (learning future)
train_X = scaler.transform(train_X)
test_X = scaler.transform(test_X)
```

```python
print(X.shape, train_X.shape, test_X.shape)
```

**Training the model**

```python
classifier = LogisticRegression(learning_rate = 0.01, number_of_iterations = 1000)
```

```python
classifier.fit(train_X, train_Y)
```

**Model Evaluation**

Accuracy Score

```python
#Accuracy score on training data
X_train_prediction = classifier.predict(train_X)
training_data_accuracy = accuracy_score(X_train_prediction, train_Y)
```

```python
print('Accuracy score on training data: ', training_data_accuracy)
```

```python
#Accuracy score on test data
X_test_prediction = classifier.predict(test_X)
test_data_accuracy = accuracy_score(X_test_prediction, test_Y)
```

```python
print('Accuracy score on test data: ', test_data_accuracy)
```
